```python
import pandas as pd 
```

```python
df = pd.read_csv(
    r"E:\Financial_Transaction_Analytics\data\cleaned_data\cleaned_data.csv",parse_dates=["transaction_datetime", "date_of_birth"]
)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df["transaction_year"] = df["transaction_datetime"].dt.year
df["transaction_month"] = df["transaction_datetime"].dt.month
df["transaction_month_name"] = df["transaction_datetime"].dt.month_name()
df["transaction_day"] = df["transaction_datetime"].dt.day
df["transaction_day_name"] = df["transaction_datetime"].dt.day_name()
df["transaction_hour"] = df["transaction_datetime"].dt.hour
```

```python
df.columns
```

```python
def get_time_of_day(hour):
    if hour>=5 and hour < 12:
        return "Morning"
    elif hour>=12 and hour<17:
        return "Afternoon"
    elif hour>=17 and hour<19:
        return "Evening"
    else:
        return "Night"
```

```python
df["time_of_day"] = df["transaction_hour"].apply(get_time_of_day)
```

```python
df.sample(20)[["transaction_hour", "time_of_day"]]
```

```python
def get_weekend(transaction_day_name):
    if transaction_day_name in ["Saturday", "Sunday"]:
        return("weekend")
    else:
        return("weekdays")
```

```python
df["weekend"] = df["transaction_day_name"].apply(get_weekend)
```

```python
df.sample(20)["weekend"]
```

```python
df["transaction_day_name"].head()
```

```python
df["customer_age"] =( (df["transaction_datetime"] - df["date_of_birth"])/pd.Timedelta(days = 365.25)).astype(int)
```

```python
df["customer_age"].head()
```

```python
bins = [0,18,35,50,65,100]
labels = ["Under 18", "18-35","35-50","50-65","65-100"]
df["customer_age_group"] = pd.cut(df["customer_age"],bins = bins, labels = labels,right = True)
```

```python
df["customer_age_group"].value_counts()
```

```python
bins = [0,50,200,500,float("inf")]

labels = ["Low",
          "Medium",
          "High",
          "Very High"]

df["transaction_amount_group"] = pd.cut(df["transaction_amount"],bins = bins, labels = labels)
```

```python
df.sample(20)["transaction_amount_group"]
```

```python
bins = [0,10000,100000,500000,float("inf")]

labels = ["Small Town",
          "Medium Town",
          "Large Town",
          "Metro"]

df["city_population_group"] = pd.cut(df["city_population"],bins = bins,labels = labels)
```

```python
df.sample(20)["city_population_group"]
```

```python
df.to_csv(r"E:\Financial_Transaction_Analytics\data\feature_engineered_data\transactions_feature_engineered.csv",index = False)
```
